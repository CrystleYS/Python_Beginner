```python
#Example dictionary

users = {
    "username": "younas",     # Username is Key and Younas is Value
    "age": 36,                # age is Key and 36 is Value
    "city": "Karachi"         # city is Key and karachi is Value
}

print(users["username"])
print(users["age"])
print(users["city"])
```

```python
# Adding and updating Pythin Dictionary

users = {
    "username": "younas",     # Username is Key and Younas is Value
    "age": 36,                # age is Key and 36 is Value
    "city": "Karachi"         # city is Key and karachi is Value
}

users ["username"] = "Umar"    # updating Username is Key and Umar is Value
users ["age"] = "8.3"          # Updating age is Key and 8.3 is Value
users ["city"] = "Bahawalpur"  # city is Key and Bahawalpur is Value


print(users["username"])
print(users["age"])
print(users["city"])


#Adding a key vlue in our dictionary
users ["country"] = "Pakistan"
```

```python
# While loop

countdown = 10

while False:
  print(f"{countdown}...")
```

```python
countdown = 10

while True:    # just start loop but not task to stop
  print(f"{countdown}...")
```

```python
# While loop

countdown = 10

# Condition

while True:
  print(f"{countdown}...")

  countdown -= 1
  print(f"{countdown}...")
```

```python
# While loop  Assignment

countdown = 10

# Condition

while countdown >= 1:
  print(f"{countdown}...")

  countdown -= 1

print("Happy New Year!!!")
```

```python
# For Loop Assignment print squqre

numbers = [1, 2, 3, 4, 5]

for num in numbers:
  print (num)
  print (num **2)
```
